# SPOT — graphical interface

Run the cells to open the desktop GUI. Select an example and quality, choose **Load example**, edit the controls, and select **Run simulation**. The GUI uses the same engine as `01_Examples.ipynb` and opens in a separate desktop window.

Choose `full` for the catalogue-based 1024 × 1024 configuration. `preview` uses reduced sampling, synthetic stars and approximate frame conversion. **Full configuration** exposes model parameters; **Edit target** exposes trajectory, material and body attitude. Plots and exports use the completed acquisition. The dataset preset produces one sample; use the examples notebook for batch generation.

For the sunlit Earth-limb case, select **space_limb → full → Load example → Run simulation**.


```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if not (ROOT / 'ssa_emulator').is_dir():
    raise RuntimeError('Start Jupyter in the python folder containing ssa_emulator.')
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import tkinter as tk
from gui import SPOTApp
print('GUI and engine:', ROOT)
```

## Launch

Run once. If the window was closed, rerun this cell. Tk event-loop integration keeps the notebook usable.

```python
%gui tk
try:
    window_open = bool(spot_window.winfo_exists())
except (NameError, tk.TclError):
    window_open = False
if not window_open:
    spot_window = tk.Tk()
    spot_app = SPOTApp(spot_window)
else:
    spot_window.lift()
```
